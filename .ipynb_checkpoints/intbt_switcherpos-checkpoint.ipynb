```python
import numpy as np
import scipy as scp
import pandas as pd
from matplotlib import pyplot as plt
from scipy.optimize import curve_fit
from scipy.stats import norm
from scipy.special import erf
from datetime import datetime
import h5py, math
```

```python
def Gauss_fit(x, A, B, y0):
    return A * np.exp(-(x)**2 / (2 * B**2)) + y0

def integrated_gauss(x, A, mu, sigma, c): # A .. aplitude, mu.. Mittelpunkt, sigma... breite, c... offset
    return c + 0.5 * A * (1 + erf((x-mu) / (np.sqrt(2) * sigma)))
```

```python
def get_data(ranges, exclude = None):
    mon = []
    stdmon = []
    data = []
    std = []
    neutrons = []
    roi0, roi1 = 90,900
    roi2, roi3 = 40, 400
    exclude = set(exclude) if exclude is not None else set()
    if isinstance(ranges, tuple):
        ranges = [ranges]
    for i0, i1 in ranges:
        for i in range(i0, i1+1):
            if i in exclude:
                continue
            path = '/home/ramann/MyData/byProposal/exp_3-14-458_262_pf2ucn/rawdata/00' + str(i) + '.nxs'
            try:
                file = h5py.File(path, 'r')
                cycletime = np.array(file['/entry0/time'][:])
                numor = np.array(file['/entry0/run_number'])
                if cycletime < 250:
                    arr1 = np.array(file['/entry0/PF2UCN/Detector1/data'][:])[roi2:roi3,0,10:210]
                    arr2 = np.array(file['/entry0/PF2UCN/Detector2/data'][:])[roi0:roi1,0,10:210]
                else:
                    neutron = np.array(file['entry0/PF2UCN/Detector2/detsum'][:])
                    print(numor, neutron, "cycletime > 250s")
                    continue      
                mon.append((np.sum(arr1) / 200)) # monitor rate in cps
                stdmon.append((math.sqrt(np.sum(arr1)) / 200)) # monitor error in cps
                data.append((np.sum(arr2) / 200)*1000) # detector rate in mcps
                std.append((math.sqrt(np.sum(arr2)) / 200)*1000) # detector error in mcps
                neutrons.append(np.sum(arr2))
            except Exception as e:
                print(e)
    return np.array(neutrons), np.array(data), np.array(std), np.array(mon), np.array(stdmon)
```

```python
def fill_data(data):
    results = []
    norm = 250

    for x, (neutrons, rate, err, mon, stdmon) in data:
        neutrons = np.array(neutrons)

        N = np.sum(neutrons)
        R = N / (200 * len(neutrons)) * 1000
        sigma = np.sqrt(N) / (200 * len(neutrons)) * 1000

        results.append({"x": x, "n": neutrons, "rate": rate, "err": err, "mon": mon, "merr": stdmon, "N": N, "R": R, "sigma": sigma})

    # Subplots: rate | monitor
    fig, axes = plt.subplots(len(results), 2, figsize=(10, 3 * len(results)), sharex=True)

    if len(results) == 1:
        axes = [axes]

    # Plotten
    for ax_row, res in zip(axes, results):
        ax_rate, ax_mon = ax_row

        y = np.asarray(res["rate"]).flatten()
        yerr = np.asarray(res["err"]).flatten()
        m = np.asarray(res["mon"]).flatten()
        merr = np.asarray(res["merr"]).flatten()

        R = res["R"]
        sigma = res["sigma"]

        y_corr = y * norm / m
        yerr_corr = y_corr * np.sqrt((yerr / y)**2 + (merr / m)**2)

        print(res['x'], len(res['n']), np.sum(res['n']), R, sigma)

        # plot rate
        ax_rate.errorbar(range(len(y)), y, yerr=yerr, fmt='o', label='rate')
        #ax_rate.errorbar(range(len(y_corr)), y_corr, yerr=yerr_corr, fmt='o', label='corrected rate', color = 'green')
        ax_rate.axhline(R, color='red', label='mean rate')
        ax_rate.axhline(R + sigma, linestyle='--', color='red', alpha=0.6)
        ax_rate.axhline(R - sigma, linestyle='--', color='red', alpha=0.6)

        ax_rate.set_title(f"x = {res['x']} (Rate)")
        ax_rate.set_ylabel("Rate [mcps]")
        ax_rate.grid(True)
        ax_rate.legend()

        # plot monitor
        ax_mon.errorbar(range(len(m)), m, yerr=merr, fmt='o', label='monitor', color='orange')

        ax_mon.set_title(f"x = {res['x']} (Monitor)")
        ax_mon.set_ylabel("Monitor [cps]")
        ax_mon.grid(True)
        ax_mon.legend()

    axes[-1][0].set_xlabel("Index")
    axes[-1][1].set_xlabel("Index")

    plt.tight_layout()
    plt.show()

    return results
```

```python
def calc_rate(data):
    x_vals = []
    rate_mean = []
    rate_err = []
    neutron_sum = []
    for x, (neutrons, rate, err, _, _) in data:
        x_vals.append(x)
    
        N = np.sum(neutrons)
        R = N / (200 * (len(neutrons+1)))*1000
        sigma = np.sqrt(N) / (200 * (len(neutrons+1)))*1000
        print("v_cut:", x, ", Neutrons:", N, ", cycles:", len(neutrons), ", rate:", R, ", sigma:", sigma, "\n")
    
        neutron_sum.append(N)
        rate_mean.append(R)
        rate_err.append(sigma)

    return np.array(x_vals), np.array(rate_mean), np.array(rate_err)
```

```python
# rate checks velocity 5-9m/s, Absorber 100/100
data100100 = []
data100100.append((0, get_data((8048,8066))))
restults = []
results = fill_data(data100100)
```
