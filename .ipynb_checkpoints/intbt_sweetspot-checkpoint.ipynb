```python
def Gauss_fit(x, A, B, y0):
    return A * np.exp(-(x)**2 / (2 * B**2)) + y0

def integrated_gauss(x, A, mu, sigma, c): # A .. aplitude, mu.. Mittelpunkt, sigma... breite, c... offset
    return c + 0.5 * A * (1 + erf((x-mu) / (np.sqrt(2) * sigma)))
```

```python
def get_data(ranges, exclude = None):
    mon = []
    stdmon = []
    data = []
    std = []
    neutrons = []
    roi0, roi1 = 90,900
    roi2, roi3 = 40, 400
    exclude = set(exclude) if exclude is not None else set()
    if isinstance(ranges, tuple):
        ranges = [ranges]
    for i0, i1 in ranges:
        for i in range(i0, i1+1):
            if i in exclude:
                continue
            path = '/home/fiscfeli/MyData/byProposal/exp_3-14-458_262_pf2ucn/rawdata/00' + str(i) + '.nxs'
            try:
                file = h5py.File(path, 'r')
                cycletime = np.array(file['/entry0/time'][:])
                numor = np.array(file['/entry0/run_number'])
                if cycletime < 250:
                    arr1 = np.array(file['/entry0/PF2UCN/Detector1/data'][:])[roi2:roi3,0,0:60]
                    arr2 = np.array(file['/entry0/PF2UCN/Detector2/data'][:])[roi0:roi1,0,0:60]
                else:
                    neutron = np.array(file['entry0/PF2UCN/Detector2/detsum'][:])
                    print(numor, neutron, "cycletime > 250s")
                    continue      
                mon.append((np.sum(arr1) / 200)) # monitor rate in cps
                stdmon.append((math.sqrt(np.sum(arr1)) / 200)) # monitor error in cps
                data.append((np.sum(arr2) / 200)*1000) # detector rate in mcps
                std.append((math.sqrt(np.sum(arr2)) / 200)*1000) # detector error in mcps
                neutrons.append(np.sum(arr2))
            except Exception as e:
                print(e)
    return np.array(neutrons), np.array(data), np.array(std), np.array(mon), np.array(stdmon)
```

```python
def calc_rate(data):
    x_vals = []
    rate_mean = []
    rate_err = []
    neutron_sum = []
    for x, (neutrons, rate, err, _, _) in data:
        x_vals.append(x)
    
        N = np.sum(neutrons)
        R = N / (200 * (len(neutrons+1)))*1000
        sigma = np.sqrt(N) / (200 * (len(neutrons+1)))*1000
        print("v_cut:", x, ", Neutrons:", N, ", cycles:", len(neutrons), ", rate:", R, ", sigma:", sigma, "\n")
    
        neutron_sum.append(N)
        rate_mean.append(R)
        rate_err.append(sigma)

    return np.array(x_vals), np.array(rate_mean), np.array(rate_err)
```

```python
# do overlap function of areas
def calc_areaoverlap(sw_mm, center, height):
    overlap = []
    k = 1.0/height
    for pos_mm in sw_mm:
        x = 1- k*abs(center-pos_mm)
        if x < 0:
            x = 0
        overlap.append(x)

    plt.plot(sw_mm, overlap, marker = "o", label='UCN')
    plt.legend()
    plt.xlabel("Switcher Motor Position [mm]")
    plt.ylabel("Relative overlap")
    plt.title("Switcher guide and turbine guide area overlap")
    plt.grid(True)
    return np.array(overlap)
```

```python
UCN_overlap = calc_areaoverlap(sw_height_002, UCN_mm, UCN_height)
```

```python
#print(get_data((8327,8350)))
def calc_areacorr_rate(overlaps):
    neutrons, rate, err, mon, stdmon = get_data((8327,8350))
    turbine_rate = np.zeros(len(overlaps))
    for i in range(len(mon)):
        if overlaps[i] > 0.01:
            turbine_rate[i] = mon[i]/overlaps[i]
    
    plt.plot(sw_height_002, turbine_rate, marker = "o", label='UCN')
    plt.legend()
    plt.xlabel("Switcher Motor Position [mm]")
    plt.ylabel("Turbine Rate [cps]")
    plt.title("Area-corrected UCN rate")
    plt.grid(True)
    return turbine_rate

area_corr_rate = calc_areacorr_rate(UCN_overlap)
```
