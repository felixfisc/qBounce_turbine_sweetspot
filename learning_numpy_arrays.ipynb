```python
def iterable(x):
    if isinstance(x, basestring): return False
    try:
        iter(x)
    except TypeError:
        return False
    return True

def dimensionality(l):
    if not iterable(l): return 0
    return 1 + dimensionality(l[0])
```

```python
# example with b = 0
a = np.array([10, 10])
b = np.array([0, 0])

try:
    # Attempt to perform the division
    result = a / b
except ZeroDivisionError:
    # Handle the case where division by zero occurs
    result = 0

# Display the result using the format method
print("Result of {} / {}: {}".format(a, b, result))
```

```python
# Configure NumPy to treat warnings as exceptions
np.seterr(all='raise')  
# This will raise exceptions for all warnings
try:
    # Code that might trigger a NumPy warning
    # For example, performing an invalid operation
    result = np.array([0, 1]) / np.array([0,0])
except Warning as e:
    print(f"Caught a NumPy warning: {e}")
except Exception as e:
    print(f"Caught an exception: {e}")
```
